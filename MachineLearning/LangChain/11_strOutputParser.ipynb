```python
import google.generativeai as genai
from dotenv import load_dotenv
import os

# Load .env file
load_dotenv()

# Configure API
genai.configure(api_key=os.getenv("GOOGLE_API_KEY"))

# List available models
for model in genai.list_models():
    
    # Show only generation models
    if "generateContent" in model.supported_generation_methods:
        print(model.name)
```

```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
import os

load_dotenv()

api_key = os.getenv("GEMINI_API_KEY")
model = ChatGoogleGenerativeAI(
    model = "gemini-2.5-flash-lite",
    google_api_key=api_key,
    temperature=0.2,
    max_output_tokens= 200
)
```

```python
from langchain_core.messages import HumanMessage

result = model.invoke([
    HumanMessage(content="How many players are in a football team? Give answer only(number)")
])

print(result.content)
```

```python
template1 = PromptTemplate(
    template="write a detetail report about {topic}",
    input_variables=["topic"]
)
template2 = PromptTemplate(
    template="give ahort 4-5 line summary on the folowing text. \n {text}",
    input_variables=["text"]
)
```

```python
prompt1 = template1.invoke({'topic' : 'Transformer'})

result = model.invoke(prompt1)

prompt2 = template2.invoke({'text' : result.content})

result = model.invoke(prompt2)

print(result.content)

```

### **Now we use output parser : -**

```python
from langchain_core.output_parsers import StrOutputParser

parser = StrOutputParser()

chain = template1 | model | parser | template2 | model | parser

result = chain.invoke({"topic" : "black Panther"})

print(result)
```
